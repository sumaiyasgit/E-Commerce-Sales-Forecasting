```python
df['order_date']=pd.to_datetime(df['order_date'])
```

```python
df['order_year'] = df['order_date'].dt.year
```

```python
df['order_month'] = df['order_date'].dt.month
df['order_day'] = df['order_date'].dt.day
df['order_month_name'] = df['order_date'].dt.month_name()
df['order_week'] = df['order_date'].dt.isocalendar().week
df['order_quarter'] = df['order_date'].dt.quarter
df['total_sales'] = df['price'] * df['quantity_sold']
df['total_sales']
```

```python
df.groupby('order_year')['total_sales'].sum()
```

```python
df.groupby('order_month')['total_sales'].sum()
```

```python
df.groupby('order_month_name')['total_sales'].sum()
```

```python
sales_by_quarter=df.groupby('order_quarter')['total_sales'].sum()
```

```python
sales_by_category=df.groupby("product_category")['total_sales'].sum()
```

```python
sales_by_product=df.groupby("product_id")['total_sales'].sum()
```

```python
sales_by_region=df.groupby("customer_region")['total_sales'].sum()
sales_by_region
```

```python
highest_selling_product=sales_by_product.sort_values(ascending=False).head(10)
highest_selling_product
```

```python
lowest_selling_product=sales_by_product.sort_values(ascending=True).head(10)
lowest_selling_product
```
